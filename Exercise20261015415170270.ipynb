習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png

import random

r = random.random()
print(f"假如 r = {r}")

binary_digits = []
temp = r
count = 0

while count < 20:
    temp *= 2
    bit = int(temp)
    binary_digits.append(str(bit))
    temp -= bit
    count += 1

binary_str = "".join(binary_digits)

print(f"r ≈ 0.{binary_str}₂")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

並查閱官方文件，了解：

- `bit_length()`  返回以二進位表示該整數所需的最小位元數
- `sys.getsizeof()`返回該 Python 物件在記憶體中佔用的實際位元組大小
- `PyLongObject`CPython 中實作整數的底層 C 結構

三者之間的關係。

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。

提示：思考利用字典，建立「數字」與「中文大寫數字」的對應關係。


def to_chinese_numeral(n):
    if n == 0:
        return "零"
    
    digits = ['零', '壹', '貳', '參', '肆', '伍', '陸', '柒', '捌', '玖']
    sub_units = ['', '拾', '佰', '仟']
    units = [
        '', '萬', '億', '兆', '京', '垓', '秭', '穰', '溝', '澗', 
        '正', '載', '極', '恆河沙', '阿僧祇', '那由他', '不可思議', '無量大數'
    ]
    
    s = str(n)
    remainder = len(s) % 4
    if remainder != 0:
        s = '0' * (4 - remainder) + s
        
    num_chunks = len(s) // 4
    chunks = [s[i*4:(i+1)*4] for i in range(num_chunks)]
    
    res = ""
    zero_needed = False
    
    for idx, chunk in enumerate(chunks):
        unit_idx = num_chunks - 1 - idx
        unit_name = units[unit_idx]
        
        chunk_res = ""
        chunk_has_nonzero = False
        
        for j, ch in enumerate(chunk):
            d = int(ch)
            sub_idx = 3 - j
            if d != 0:
                if zero_needed and chunk_res != "":
                    chunk_res += digits[0]
                    zero_needed = False
                chunk_res += digits[d] + sub_units[sub_idx]
                chunk_has_nonzero = True
            else:
                if chunk_res != "":
                    zero_needed = True
                    
        if chunk_has_nonzero:
            if chunk[0] == '0' and res != "" and not res.endswith('零') and not chunk_res.startswith('零'):
                chunk_res = digits[0] + chunk_res
            res += chunk_res + unit_name
            zero_needed = False
        else:
            if res != "" and idx < num_chunks - 1:
                if any(int(c) > 0 for c in chunks[idx+1:]):
                    zero_needed = True
                    
    return res
