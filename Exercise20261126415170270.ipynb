某線上購物平台需要計算商品的最終價格。計算時必須考慮：    
- `price`：商品原價。                                     
- `discount`：折扣率。                                    
- `tax`：稅率。                                           

價格計算公式：
$$                                        
\text{price}\times(1-\text{discount})\times(1+\text{tax})\
$$                                                        
                                                          
例如：                                                    
                                                          
- 商品原價：1000 元                                       
- 折扣率：20%                                             
- 稅率：5%                                                
                                                          
價格為：                                              
                                                         
$$                                                       
1000\times(1-0.2)\times(1+0.05)=840\                      
$$                                      
                                                          
                                                          

````python
def final_price(price, discount, tax):       
    return price * (1 - discount) * (1 + tax)                        

print(final_price(1000, 0.2, 0.05))          
````

某線上購物平台需要計算商品的最終價格，計算公式如下：

**請完成以下小題：**

1. 將原始函數改寫成 Currying（柯里化）形式，使函數呼叫方式變成：
   ```
   final_price_curried(discount)(tax)(price)
   ```
2. 使用 `lambda` expression 改寫上述柯里化函數。
3. 假設折扣率為 20%，稅率為 5%，建立一個新的函數 `calculate`，使得只需要輸入商品價格，就能計算最終價格。
4. 使用 `calculate` 計算原價為 1000、2000、3000 元的商品價格。


這題有三個觀念：

| 概念                    | 說明                |
| --------------------- | ----------------- |
| Higher-Order Function | 函數可以回傳另一個函數       |
| Currying              | 將多參數函數轉換成一連串單參數函數 |
| Partial Application   | 預先固定部分參數，產生新的函數   |

注意：**Currying 與 Partial Application 並不是同一件事。**

- Currying：改變函數的參數結構。
- Partial Application：固定部分參數，產生新的函數。

````python
# 作答區 1.
def final_price_curried(discount):
    def with_tax(tax):
        def with_price(price):
            return price * (1 - discount) * (1 + tax)
        return with_price
    return with_tax
````

````python
# 作答區 2.
final_price_curried = lambda discount: lambda tax: lambda price: price * (1 - discount) * (1 + tax)
````

````python
# 作答區 3.
calculate = final_price_curried(0.2)(0.05)
````

````python
# 作答區 4.
print(calculate(1000))
print(calculate(2000))
print(calculate(3000))
````
